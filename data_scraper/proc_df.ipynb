```python
def filter_company(df: pd.DataFrame, c_loc: str, *c_name_misc: str):
	tgt_headers = df['title'].str.lower().str.split(" ")
	req_cols = ["title","company_number","company_status","address.country"]
	mask = tgt_headers.apply(lambda words: any(arg in words for arg in c_name_misc))
	filtered_df = df[req_cols][mask]
	print(filtered_df.head())
	if c_loc:
		filtered_df = filtered_df[filtered_df["address.country"].str.lower() == c_loc.strip().lower()]
	# print(filtered_df.head())
	return filtered_df, df[req_cols]
```

```python
filt_df, all_df = filter_company(search_df,comp_name, *args)
```
